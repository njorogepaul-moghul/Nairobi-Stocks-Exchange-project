```python
import pandas as pd
import glob
import os
```

```python
# Path to our CSVs
csv_folder = r"C:\Users\Win\NSE Project\Data" # adjust if different
all_files = glob.glob(os.path.join(csv_folder, "NSE_data_all_stocks_*.csv"))
```

```python
# Check columns in the first file
df = pd.read_csv(all_files[1])
print(df.columns)
```

```python

df_list = []
for file in all_files:
    df = pd.read_csv(file)
    df.columns = df.columns.str.strip()  # remove any extra spaces
    
    # Convert DATE with explicit format
    df['DATE'] = pd.to_datetime(df['DATE'], format='%d-%b-%y', errors='coerce')
    
    df_list.append(df)

# Merge all CSVs into a single DataFrame
nse_data = pd.concat(df_list, ignore_index=True)
```

```python
# now we filter our SCOM and KCB CODE (tickers)
#for safaricom
scom=nse_data[nse_data["CODE"]== "SCOM"].sort_values("DATE")
kcb=nse_data[nse_data["CODE"]=="KCB"].sort_values("DATE")
```
