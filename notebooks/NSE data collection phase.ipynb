## Phase 1: Data Collection & Infrastructure

The first step in building a predictive dashboard is acquiring and organizing reliable data. For this project, we focus on **Nairobi Securities Exchange (NSE) stock data**, providing a real-world context relevant to local financial markets.

We source historical stock prices from a **comprehensive dataset (2007–2024) of all NSE-listed companies**, available via Mendeley Data. Selected tickers include **Safaricom (SCOM)** and **KCB (KCB)**, two of the most actively traded stocks in Kenya.

The goal of this phase is to create a **clean, structured dataset** suitable for exploratory data analysis, feature engineering, and predictive modeling. Key considerations include:

- **Completeness:** Ensuring all trading days are represented.  
- **Accuracy:** Verifying that prices are correctly recorded for each day.  
- **Format:** Organizing data in a tabular structure with timestamps as the index for easy time series analysis.

By establishing a solid data foundation from this dataset, we ensure that subsequent analysis and modeling are **reliable, reproducible, and meaningful** for both visualization and forecasting.

```python
# lets import our libraries
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Win\NSE Project\NSE_data_all_stocks_2024.csv")
```

```python
#lets convert Date column to datetime and parse the dates safely
df['Date'] = pd.to_datetime(df['Date'], format="%Y-%m-%d", errors='coerce')
```

```python
#lets see how many unique tickers we have
df["Code"].unique()
```

```python
# now we filter our SCOM and KCB tickers
#for safaricom
scom=df[df["Code"]== "SCOM"].sort_values("Date")
kcb=df[df["Code"]=="KCB"].sort_values("Date")
```
