## Phase 1: Data Collection & Infrastructure

The first stage of this project focuses on acquiring, organizing, and preparing reliable financial data from the Nairobi Securities Exchange (NSE). A robust data foundation is essential for any predictive analytics system, as the quality and structure of the dataset directly influence the reliability of downstream analysis and forecasting models.

For this project, historical NSE stock market data was sourced from publicly available datasets covering multiple years of trading activity. The datasets include key market variables such as trading date, stock code, company name, daily price information, and trading volume. These datasets were downloaded as CSV files and stored within the project directory for processing.

Using Python and the pandas library, the datasets were programmatically loaded and consolidated into a single structured DataFrame. During this step, several preprocessing tasks were performed to ensure consistency across files, including standardizing column names, handling inconsistent formats, and converting the trading date column into a proper datetime format (`%d-%b-%y`). This step ensures that the dataset can be used effectively for time series analysis.

Once the data was successfully integrated, the focus shifted to identifying relevant securities for analysis. Two of the most actively traded and economically significant stocks on the NSE were selected:

- **Safaricom PLC (SCOM)**
- **KCB Group PLC (KCB)**

These stocks provide an excellent case study due to their liquidity, market relevance, and availability of historical trading records.

The output of this phase is a clean, consolidated dataset containing NSE trading data, which will serve as the foundation for subsequent stages of the project, including exploratory data analysis (EDA), feature engineering, predictive modeling, and the development of an interactive forecasting dashboard.

```python
import pandas as pd
import glob
import os
```

```python
# Path to our CSVs
csv_folder = r"C:\Users\Win\NSE Project\Data" 
all_files = glob.glob(os.path.join(csv_folder, "NSE_data_all_stocks_*.csv"))
```

```python
# Check columns in the first file
df = pd.read_csv(all_files[1])
print(df.columns)
```

```python

df_list = []
for file in all_files:
    df = pd.read_csv(file)
    df.columns = df.columns.str.strip()  # remove any extra spaces
    
    # Convert DATE with explicit format
    df['DATE'] = pd.to_datetime(df['DATE'], format='%d-%b-%y', errors='coerce')
    
    df_list.append(df)

# Merge all CSVs into a single DataFrame
nse_data = pd.concat(df_list, ignore_index=True)
```

```python
# now we filter our SCOM and KCB CODE (tickers)
#for safaricom
scom=nse_data[nse_data["CODE"]== "SCOM"].sort_values("DATE")
kcb=nse_data[nse_data["CODE"]=="KCB"].sort_values("DATE")
```

```python
# 1. Remove the '%' symbol
# 2. Replace the dash '-' (used for no change) with '0'
# 3. Remove commas (if any)
scom['Change%'] = scom['Change%'].astype(str).str.replace('%', '', regex=False)
scom['Change%'] = scom['Change%'].replace('-', '0').str.replace(',', '', regex=False)
scom['Change'] = scom['Change'].replace('-', '0').str.replace(',', '', regex=False)

# 4. Now convert to float safely
scom['Change%'] = pd.to_numeric(scom['Change%'], errors='coerce')
scom['Change'] = pd.to_numeric(scom['Change'], errors='coerce')

# 5. Convert to decimal (0.02 becomes 0.0002) for mathematical modeling
scom['Change%'] = scom['Change%'] / 100
```

```python
# 1. Remove the '%' symbol
# 2. Replace the dash '-' (used for no change) with '0'
# 3. Remove commas (if any)
kcb['Change%'] = kcb['Change%'].astype(str).str.replace('%', '', regex=False)
kcb['Change%'] = kcb['Change%'].replace('-', '0').str.replace(',', '', regex=False)
kcb['Change'] = kcb['Change'].replace('-', '0').str.replace(',', '', regex=False)
# 4. Now convert to float safely
kcb['Change%'] = pd.to_numeric(scom['Change%'], errors='coerce')
kcb['Change'] = pd.to_numeric(scom['Change'], errors='coerce')

# 5. Convert to decimal (0.02 becomes 0.0002) for mathematical modeling
kcb['Change%'] = scom['Change%'] / 100
```

```python
# we now convert the datatypes to the correct ones
#for safaricom data:
cols=[ '12m Low', '12m High', 'Day Low', 'Day High','Day Price', 'Previous', 'Volume', 'Adjust']
for c in cols:
    scom[c]=pd.to_numeric(scom[c].str.replace(',',''),errors= 'coerce')
```
